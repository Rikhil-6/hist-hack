```python
# 1️⃣ Load environment variables
from dotenv import load_dotenv
import os

load_dotenv()  # Reads the .env file

# 2️⃣ Get the API key from environment
api_key = os.getenv("GOOGLE_API_KEY")
```

```python
pdf_name = "CO273-15"
```

```python
import fitz  # PyMuPDF
import os
from PIL import Image
from google import genai

# -----------------------------------
# 1️⃣ Initialize the GenAI Client
# -----------------------------------
# You can set GEMINI_API_KEY as an environment variable,
# or pass it explicitly below.
client = genai.Client(api_key=api_key)

# -----------------------------------
# 2️⃣ Extract PDF Pages to Images
# -----------------------------------
pdf_path = f"../data/{pdf_name}.pdf"
#pdf_name = os.path.splitext(os.path.basename(pdf_path))[0]
output_folder = "../data/pdf_images"
os.makedirs(output_folder, exist_ok=True)

doc = fitz.open(pdf_path)
image_paths = []

#for i in range(len(doc)):
for i in range(4,5):
    page = doc[i]
    pix = page.get_pixmap(dpi=300)
    path = f"{output_folder}/{pdf_name}_page_{i+1}.png"
    pix.save(path)
    image_paths.append(path)

doc.close()
```

```python
import glob
from google import genai
from google.genai.types import Part

# Initialize client
client = genai.Client(api_key=api_key)
model_name = "gemini-2.5-flash"

# Get all matching files (sorted numerically)
image_paths = sorted(
    glob.glob(f"../data/pdf_images/{pdf_name}_page_*.png"),
    key=lambda x: int(x.split("_")[-1].split(".")[0])
)

all_text = ""

for image_path in image_paths:
    print(f"Processing: {image_path}")

    with open(image_path, "rb") as f:
        img_bytes = f.read()

    response = client.models.generate_content(
        model=model_name,
        contents=[
            "Extract all handwritten text from this image. "
            "Preserve line breaks. Only output the extracted text. "
            "Refine it to accurate, readable English.",
            Part.from_bytes(data=img_bytes, mime_type="image/png"),
        ],
    )

    page_text = response.text
    all_text += f"\n\n--- {image_path} ---\n{page_text}"

print("\n\n===== FINAL OUTPUT =====\n")
print(all_text)
```
